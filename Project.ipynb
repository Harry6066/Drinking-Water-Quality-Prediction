```python
# ============================================================
# Drinking Water Quality Prediction Using Machine Learning
# Part 1 : Data Preprocessing & Visualization
# ============================================================

# Import Libraries
import os
import sys
import pickle
import pandas as pd
import matplotlib.pyplot as plt

# Create folder to save graphs
os.makedirs("graphs", exist_ok=True)

# Save all output to a text file
output_file = open("project_output.txt", "w")
sys.stdout = output_file

print("=" * 60)
print("DRINKING WATER QUALITY PREDICTION USING MACHINE LEARNING")
print("=" * 60)

# ============================================================
# Step 1 : Load Dataset
# ============================================================

print("\nStep 1 : Loading Dataset")

df = pd.read_csv("water_potability.csv")

print("\nFirst 5 Rows")
print(df.head())

print("\nDataset Information")
df.info()

print("\nDataset Shape")
print(df.shape)

print("\nColumn Names")
print(df.columns)

# ============================================================
# Step 2 : Check Missing Values
# ============================================================

print("\nStep 2 : Missing Values Before Preprocessing")

print(df.isnull().sum())

# Missing Values Graph

plt.figure(figsize=(8,5))

df.isnull().sum().plot(
    kind="bar",
    color="orange"
)

plt.title("Missing Values Before Preprocessing")
plt.xlabel("Features")
plt.ylabel("Missing Values")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("graphs/missing_values.png")

plt.close()

print("\nMissing Values Graph Saved")

# Fill Missing Values

df.fillna(df.mean(numeric_only=True), inplace=True)

print("\nMissing Values After Preprocessing")

print(df.isnull().sum())

# ============================================================
# Step 3 : Remove Duplicate Rows
# ============================================================

duplicates = df.duplicated().sum()

print("\nDuplicate Rows :", duplicates)

df.drop_duplicates(inplace=True)

print("\nDataset Shape After Removing Duplicates")

print(df.shape)

# ============================================================
# Step 4 : Save Cleaned Dataset
# ============================================================

df.to_csv(
    "water_potability_cleaned.csv",
    index=False
)

print("\nCleaned Dataset Saved Successfully")

# ============================================================
# Step 5 : Class Distribution
# ============================================================

print("\nCreating Class Distribution Graph")

plt.figure(figsize=(6,4))

df["Potability"].value_counts().plot(
    kind="bar",
    color=["red","green"]
)

plt.title("Water Potability Distribution")
plt.xlabel("Potability")
plt.ylabel("Count")

plt.xticks(
    [0,1],
    ["Unsafe","Safe"],
    rotation=0
)

plt.tight_layout()

plt.savefig("graphs/class_distribution.png")

plt.close()

print("Class Distribution Graph Saved")

# ============================================================
# Step 6 : Correlation Heatmap
# ============================================================

print("\nCreating Correlation Heatmap")

plt.figure(figsize=(10,8))

plt.imshow(
    df.corr(),
    cmap="coolwarm"
)

plt.colorbar()

plt.xticks(
    range(len(df.columns)),
    df.columns,
    rotation=90
)

plt.yticks(
    range(len(df.columns)),
    df.columns
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig("graphs/correlation_heatmap.png")

plt.close()

print("Correlation Heatmap Saved")

# ============================================================
# Step 7 : Feature Selection
# ============================================================

print("\nSplitting Features and Target")

X = df.drop("Potability", axis=1)

y = df["Potability"]

print("Features Shape :", X.shape)
print("Target Shape   :", y.shape)

print("\nData Preprocessing Completed Successfully")

# ============================================================
# End of Part 1
# ============================================================
```

```python
# ============================================================
# Part 2 : Model Training & Evaluation
# ============================================================

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

# ============================================================
# Step 8 : Split Dataset
# ============================================================

print("\nStep 8 : Splitting Dataset")

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data Shape :", X_train.shape)
print("Testing Data Shape  :", X_test.shape)

print("\nTraining Samples :", len(X_train))
print("Testing Samples  :", len(X_test))

# ============================================================
# Step 9 : Train Random Forest Model
# ============================================================

print("\nStep 9 : Training Random Forest Model")

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

print("Model Training Completed Successfully")

# ============================================================
# Step 10 : Prediction
# ============================================================

print("\nStep 10 : Predicting Test Data")

y_pred = model.predict(X_test)

# ============================================================
# Step 11 : Model Evaluation
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

print("\n" + "=" * 60)
print("MODEL EVALUATION")
print("=" * 60)

print(f"\nModel Accuracy : {accuracy*100:.2f}%")

print("\nConfusion Matrix")
cm = confusion_matrix(y_test, y_pred)
print(cm)

print("\nClassification Report")
print(classification_report(y_test, y_pred))

# ============================================================
# Step 12 : Accuracy Graph
# ============================================================

print("\nCreating Accuracy Graph")

plt.figure(figsize=(6,5))

plt.bar(
    ["Random Forest"],
    [accuracy * 100],
    color="green"
)

plt.title("Random Forest Accuracy")
plt.ylabel("Accuracy (%)")
plt.ylim(0,100)

plt.text(
    0,
    accuracy*100 + 1,
    f"{accuracy*100:.2f}%",
    ha="center",
    fontsize=12
)

plt.tight_layout()

plt.savefig("graphs/accuracy_chart.png")

plt.close()

print("Accuracy Graph Saved")

# ============================================================
# Step 13 : Confusion Matrix Heatmap
# ============================================================

print("\nCreating Confusion Matrix Heatmap")

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Unsafe","Safe"]
)

disp.plot(cmap="Blues")

plt.title("Confusion Matrix")

plt.tight_layout()

plt.savefig("graphs/confusion_matrix.png")

plt.close()

print("Confusion Matrix Graph Saved")
```

```python
# ============================================================
# Part 3 : Feature Importance, Save Model & Finish Program
# ============================================================

# ============================================================
# Step 14 : Feature Importance Graph
# ============================================================

print("\nStep 14 : Creating Feature Importance Graph")

importance = pd.Series(
    model.feature_importances_,
    index=X.columns
)

importance = importance.sort_values(ascending=True)

plt.figure(figsize=(8,6))

importance.plot(
    kind="barh",
    color="skyblue"
)

plt.title("Feature Importance")
plt.xlabel("Importance Score")
plt.ylabel("Features")

plt.tight_layout()

plt.savefig("graphs/feature_importance.png")

plt.close()

print("Feature Importance Graph Saved")

# ============================================================
# Step 15 : Save Trained Model
# ============================================================

print("\nStep 15 : Saving Trained Model")

with open("water_model.pkl", "wb") as file:
    pickle.dump(model, file)

print("Model saved successfully as 'water_model.pkl'")

# ============================================================
# Step 16 : Project Summary
# ============================================================

print("\n" + "=" * 60)
print("PROJECT SUMMARY")
print("=" * 60)

print(f"Dataset Shape              : {df.shape}")
print(f"Training Samples           : {len(X_train)}")
print(f"Testing Samples            : {len(X_test)}")
print(f"Model Used                 : Random Forest")
print(f"Model Accuracy             : {accuracy*100:.2f}%")

print("\nGenerated Files:")
print("1. water_potability_cleaned.csv")
print("2. water_model.pkl")
print("3. project_output.txt")

print("\nGenerated Graphs:")
print("1. missing_values.png")
print("2. class_distribution.png")
print("3. correlation_heatmap.png")
print("4. accuracy_chart.png")
print("5. confusion_matrix.png")
print("6. feature_importance.png")

print("\nProject Completed Successfully!")

print("=" * 60)
print("END OF PROGRAM")
print("=" * 60)

# ============================================================
# Close Output File
# ============================================================

sys.stdout.close()
sys.stdout = sys.__stdout__

print("Project executed successfully!")
print("Complete output saved in 'project_output.txt'")
print("Graphs saved inside the 'graphs' folder.")
```
